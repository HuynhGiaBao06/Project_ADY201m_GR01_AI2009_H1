# CatBoost Additional Benchmark
Uses the active ordinal-encoded tree branch without `cat_features` to preserve benchmark consistency.

## 1. Imports, configuration, and output setup

```python
from __future__ import annotations
from pathlib import Path
import sys
import warnings
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from IPython.display import display
from catboost import CatBoostClassifier
from sklearn.metrics import (accuracy_score, average_precision_score, classification_report, confusion_matrix, f1_score, precision_recall_curve, precision_score, recall_score, roc_auc_score, roc_curve)
cwd = Path.cwd().resolve()
if (cwd / 'Source' / 'src').is_dir(): REPO_ROOT = cwd
elif cwd.name == 'notebooks' and cwd.parent.name == 'Source': REPO_ROOT = cwd.parent.parent
else: raise RuntimeError(f'Cannot resolve repository root from {cwd}')
if str(REPO_ROOT) not in sys.path: sys.path.insert(0, str(REPO_ROOT))
from Source.src.path import DATA_REPORTS_DIR, FIGURES_DIR, TRAIN_DATA_FILE, VALIDATION_DATA_FILE, TEST_DATA_FILE
from Source.src.utils.feature_display import build_feature_mapping_table
warnings.filterwarnings('default')
RANDOM_STATE=42
DEFAULT_THRESHOLD=0.50
TARGET_RECALL=0.85
F1_TOLERANCE=1e-6
THRESHOLDS=np.round(np.arange(0.05,0.951,0.01),2)
TOP_N_FEATURES=15
MODEL_NAME='CatBoost'
assert len(THRESHOLDS)==91
required=[DATA_REPORTS_DIR,FIGURES_DIR,TRAIN_DATA_FILE,VALIDATION_DATA_FILE,TEST_DATA_FILE]
missing=[p for p in required if not p.exists()]
if missing: raise FileNotFoundError(f'Missing required paths; stopping before run_pipeline: {missing}')
print('Repository root:',REPO_ROOT)
print('Model:',MODEL_NAME,'threshold candidates:',len(THRESHOLDS))
print('All checkpoints exist; ingestion must remain skipped.')
```

## 2. Evaluation and export helpers

```python
def save_table(df,filename):
    path=DATA_REPORTS_DIR/filename; df.to_csv(path,index=False); print(f'Saved {filename}: shape={df.shape}'); return path
def save_figure(filename):
    path=FIGURES_DIR/filename; plt.tight_layout(); plt.savefig(path,dpi=300,bbox_inches='tight'); print('Saved 300-DPI figure:',path); plt.show(); plt.close(); return path
def evaluate(y_true,y_proba,threshold,split):
    y_true=np.asarray(y_true).astype(int); y_proba=np.asarray(y_proba); y_pred=(y_proba>=threshold).astype(int)
    tn,fp,fn,tp=confusion_matrix(y_true,y_pred,labels=[0,1]).ravel()
    return pd.DataFrame([{'Model':MODEL_NAME,'Split':split,'Threshold':float(threshold),'Accuracy':accuracy_score(y_true,y_pred),'Precision':precision_score(y_true,y_pred,zero_division=0),'Recall':recall_score(y_true,y_pred,zero_division=0),'F1_Score':f1_score(y_true,y_pred,zero_division=0),'ROC_AUC':roc_auc_score(y_true,y_proba),'PR_AUC':average_precision_score(y_true,y_proba),'TP':int(tp),'FP':int(fp),'TN':int(tn),'FN':int(fn),'False_Positive_Rate':fp/(fp+tn) if fp+tn else 0.0,'False_Negative_Rate':fn/(fn+tp) if fn+tp else 0.0}])
def scan_thresholds(y_true,y_proba):
    table=pd.concat([evaluate(y_true,y_proba,t,'validation') for t in THRESHOLDS],ignore_index=True); table['Meets_Recall_Target']=table['Recall']>=TARGET_RECALL
    if len(table)!=91: raise RuntimeError(f'Expected 91 threshold rows, got {len(table)}')
    return table
def select_threshold(table):
    eligible=table[table['Recall']>=TARGET_RECALL].copy()
    if not eligible.empty:
        max_f1=eligible['F1_Score'].max(); tied=eligible[eligible['F1_Score']>=max_f1-F1_TOLERANCE]; selected=tied.sort_values(['Precision','Threshold'],ascending=[False,False]).iloc[0]
        return float(selected['Threshold']),True,'Recall target met; highest F1 with 1e-6 tolerance, then higher precision and threshold.'
    selected=table.sort_values(['Recall','F1_Score','Precision','Threshold'],ascending=[False]*4).iloc[0]
    return float(selected['Threshold']),False,'Recall target not met; highest recall, then F1, precision, and threshold.'
def confusion_table(y_true,y_proba,threshold,split):
    matrix=confusion_matrix(y_true,(np.asarray(y_proba)>=threshold).astype(int),labels=[0,1])
    return pd.DataFrame([{'Split':split,'Threshold':threshold,'Actual':a,'Predicted':p,'Count':int(matrix[i,j])} for i,a in enumerate([0,1]) for j,p in enumerate([0,1])])
def report_table(y_true,y_proba,threshold,split):
    pred=(np.asarray(y_proba)>=threshold).astype(int); report=pd.DataFrame(classification_report(y_true,pred,labels=[0,1],output_dict=True,zero_division=0)).T.reset_index().rename(columns={'index':'Class'}); report.insert(0,'Threshold',threshold); report.insert(0,'Split',split); return report
def plot_confusion(y_true,y_proba,threshold,title):
    matrix=confusion_matrix(y_true,(np.asarray(y_proba)>=threshold).astype(int),labels=[0,1]); fig,ax=plt.subplots(figsize=(6,5)); image=ax.imshow(matrix,cmap='Blues'); ax.set(title=title,xlabel='Predicted label',ylabel='Actual label',xticks=[0,1],yticks=[0,1]); ax.set_xticklabels(['No Churn','Churn']); ax.set_yticklabels(['No Churn','Churn'])
    for i in range(2):
        for j in range(2): ax.text(j,i,int(matrix[i,j]),ha='center',va='center',color='white' if matrix[i,j]>matrix.max()/2 else 'black')
    fig.colorbar(image,ax=ax,fraction=0.046,pad=0.04)
```

## 3. Load active tree pipeline outputs

```python
from Source.src.main import run_pipeline
result=run_pipeline(force_ingestion=False,use_database=False)
builder=result['builder']; outputs=result['outputs']
tree_train=outputs['tree']['train']; tree_val=outputs['tree']['validation']; tree_test=outputs['tree']['test']
print("Pipeline branch used: outputs['tree'] only")
print('Tree shapes:',tree_train.shape,tree_val.shape,tree_test.shape)
```

## 4. Validate split and feature contract

```python
expected_tree={'train':(4930,20),'validation':(1056,20),'test':(1057,20)}
for name,frame in {'train':tree_train,'validation':tree_val,'test':tree_test}.items():
    assert frame.shape==expected_tree[name]; assert {'ID','Churn'}.issubset(frame.columns)
X_train=tree_train.drop(columns=['ID','Churn']); y_train=tree_train['Churn'].astype(int)
X_val=tree_val.drop(columns=['ID','Churn']); y_val=tree_val['Churn'].astype(int)
X_test=tree_test.drop(columns=['ID','Churn']); y_test=tree_test['Churn'].astype(int)
expected_x={'train':(4930,18),'validation':(1056,18),'test':(1057,18)}
for name,frame in {'train':X_train,'validation':X_val,'test':X_test}.items():
    assert frame.shape==expected_x[name]; assert list(frame.columns)==list(X_train.columns); assert 'ID' not in frame and 'Churn' not in frame
    assert not frame.select_dtypes(include=['object','category']).columns.tolist(); assert not frame.isna().any().any(); assert np.isfinite(frame.to_numpy(dtype=float)).all()
for target in [y_train,y_val,y_test]: assert set(target.unique()).issubset({0,1}) and target.notna().all()
print('X/y shapes:',X_train.shape,y_train.shape,X_val.shape,y_val.shape,X_test.shape,y_test.shape)
print('Feature columns:',X_train.columns.tolist())
print('No raw CatBoost categorical handling and no cat_features are used.')
print('Training class distribution:'); display(y_train.value_counts().sort_index().rename('count'))
```

## 5. Model configuration and training

```python
catboost_model=CatBoostClassifier(iterations=300,learning_rate=0.05,depth=6,loss_function='Logloss',eval_metric='AUC',auto_class_weights='Balanced',random_seed=RANDOM_STATE,verbose=False,allow_writing_files=False)
print('Model parameters:',catboost_model.get_params())
catboost_model.fit(X_train,y_train)
validation_probability=catboost_model.predict_proba(X_val)[:,1]
print('Training completed; validation probability shape:',validation_probability.shape)
```

## 6. Default-threshold validation evaluation

```python
default_metrics=evaluate(y_val,validation_probability,DEFAULT_THRESHOLD,'validation'); display(default_metrics)
save_table(default_metrics,'catboost_default_threshold_validation_metrics.csv')
save_table(confusion_table(y_val,validation_probability,DEFAULT_THRESHOLD,'validation'),'catboost_validation_confusion_matrix_default_threshold.csv')
save_table(report_table(y_val,validation_probability,DEFAULT_THRESHOLD,'validation'),'catboost_validation_classification_report_default_threshold.csv')
plot_confusion(y_val,validation_probability,DEFAULT_THRESHOLD,'CatBoost Validation Confusion Matrix - Threshold 0.50'); save_figure('catboost_01_validation_confusion_matrix_default_threshold.png')
```

## 7. Threshold Sensitivity Analysis

```python
threshold_table=scan_thresholds(y_val,validation_probability); save_table(threshold_table,'catboost_threshold_sensitivity_validation.csv')
selected_threshold,recall_target_met,selection_rule=select_threshold(threshold_table)
selected_validation=evaluate(y_val,validation_probability,selected_threshold,'validation')
selected_validation['Meets_Recall_Target']=selected_validation['Recall']>=TARGET_RECALL; selected_validation['Target_Recall']=TARGET_RECALL; selected_validation['Recall_Target_Met']=recall_target_met; selected_validation['Selection_Rule']=selection_rule
display(selected_validation); print('Selected validation threshold:',selected_threshold)
save_table(selected_validation,'catboost_selected_threshold_validation_metrics.csv')
save_table(confusion_table(y_val,validation_probability,selected_threshold,'validation'),'catboost_validation_confusion_matrix_selected_threshold.csv')
save_table(report_table(y_val,validation_probability,selected_threshold,'validation'),'catboost_validation_classification_report_selected_threshold.csv')
fig,ax=plt.subplots(figsize=(10,6))
for metric in ['Precision','Recall','F1_Score']: ax.plot(threshold_table['Threshold'],threshold_table[metric],label=metric)
ax.axhline(TARGET_RECALL,color='gray',linestyle='--',label='Recall target 0.85'); ax.axvline(selected_threshold,color='black',linestyle=':',label=f'Selected {selected_threshold:.2f}'); ax.set(title='CatBoost Threshold Sensitivity Analysis - Validation',xlabel='Threshold',ylabel='Score',ylim=(0,1.05)); ax.grid(alpha=.3); ax.legend(); save_figure('catboost_02_threshold_sensitivity_validation.png')
```

## 8. Validation ROC and precision-recall curves

```python
fpr,tpr,_=roc_curve(y_val,validation_probability); plt.figure(figsize=(7,6)); plt.plot(fpr,tpr,label=f'ROC-AUC = {roc_auc_score(y_val,validation_probability):.4f}'); plt.plot([0,1],[0,1],'--',color='gray'); plt.xlabel('False Positive Rate'); plt.ylabel('True Positive Rate'); plt.title('CatBoost Validation ROC Curve'); plt.grid(alpha=.3); plt.legend(); save_figure('catboost_03_roc_curve_validation.png')
precision,recall,_=precision_recall_curve(y_val,validation_probability); plt.figure(figsize=(7,6)); plt.plot(recall,precision,label=f'PR-AUC = {average_precision_score(y_val,validation_probability):.4f}'); plt.xlabel('Recall'); plt.ylabel('Precision'); plt.title('CatBoost Validation Precision-Recall Curve'); plt.grid(alpha=.3); plt.legend(); save_figure('catboost_04_precision_recall_curve_validation.png')
```

## 9. Final frozen-threshold test evaluation

```python
test_probability=catboost_model.predict_proba(X_test)[:,1]
test_metrics=evaluate(y_test,test_probability,selected_threshold,'test'); test_metrics['Threshold_Source']='validation'; display(test_metrics)
save_table(test_metrics,'catboost_selected_threshold_test_metrics.csv')
save_table(confusion_table(y_test,test_probability,selected_threshold,'test'),'catboost_test_confusion_matrix_selected_threshold.csv')
save_table(report_table(y_test,test_probability,selected_threshold,'test'),'catboost_test_classification_report_selected_threshold.csv')
plot_confusion(y_test,test_probability,selected_threshold,f'CatBoost Test Confusion Matrix - Frozen Threshold {selected_threshold:.2f}'); save_figure('catboost_05_test_confusion_matrix_selected_threshold.png')
```

## 10. Native feature importance

```python
mapping=build_feature_mapping_table(X_train.columns); save_table(mapping,'feature_name_mapping_catboost.csv')
importance=mapping.copy(); importance.insert(0,'Model',MODEL_NAME); importance['Importance_Type']='CatBoost FeatureImportance'; importance['Importance']=catboost_model.get_feature_importance(type='FeatureImportance').astype(float)
importance=importance.sort_values('Importance',ascending=False).reset_index(drop=True); importance['Rank']=np.arange(1,len(importance)+1)
save_table(importance,'catboost_feature_importance.csv'); top_importance=importance.head(TOP_N_FEATURES).copy(); save_table(top_importance,'catboost_top_feature_importance.csv')
display(top_importance[['Rank','Feature_Display_Name','Importance']])
plot_df=top_importance.iloc[::-1]; plt.figure(figsize=(10,7)); plt.barh(plot_df['Feature_Display_Name'],plot_df['Importance']); plt.xlabel('CatBoost FeatureImportance'); plt.title('CatBoost Top Native Feature Importance'); plt.grid(axis='x',alpha=.3); save_figure('catboost_06_top_feature_importance.png')
cumulative=importance['Importance'].cumsum()/importance['Importance'].sum(); plt.figure(figsize=(9,6)); plt.plot(importance['Rank'],cumulative,marker='o'); plt.axhline(.8,color='gray',linestyle='--'); plt.xlabel('Feature rank'); plt.ylabel('Cumulative importance share'); plt.title('CatBoost Cumulative Feature Importance'); plt.grid(alpha=.3); save_figure('catboost_07_cumulative_feature_importance.png')
```
